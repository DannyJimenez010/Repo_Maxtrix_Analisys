Este código permite generar n matrices de tamaño 4x4 bajo una condición lógica seleccionada:

- Fibonacci
- Múltiplos de 5

Las matrices se apilan una debajo de otra automáticamente.

```python
import numpy as np

def generar(n, tipo):
    
    for k in range(n):
        
        if tipo == 1:  # Fibonacci
            a, b = np.random.randint(1,21,2)
            fib = [a, b]
            for i in range(14):
                fib.append(fib[-1] + fib[-2])
            m = np.array(fib).reshape(4,4)
            nombre = "Fibonacci"
        
        else:  # Multiplos de 5
            inicio = np.random.randint(1,20)
            m = np.arange(inicio, inicio + 5*16, 5).reshape(4,4)
            nombre = "Multiplos de 5"
        
        print(f"\nMatriz {k+1} ({nombre})")
        print(m)


n = int(input("Ingrese número de matrices: "))
tipo = int(input("1 Fibonacci | 2 Multiplos de 5: "))

print(f"\nGenerando {n} matrices de 4x4\n")

generar(n, tipo)
```
